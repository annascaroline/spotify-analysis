```python
# Cell 3 — Pull tracks via search (limit fixed to 20)
import pandas as pd

def get_tracks_by_genre(search_term, genre_label):
    tracks = []
    results = sp.search(q=search_term, type='track', limit=20)
    for item in results['tracks']['items']:
        if item is None or item.get('id') is None:
            continue
        tracks.append({
            'id':     item['id'],
            'name':   item['name'],
            'artist': item['artists'][0]['name'],
            'genre':  genre_label
        })
    return tracks

all_tracks = []
all_tracks.extend(get_tracks_by_genre("pop hits 2024",       "Pop"))
all_tracks.extend(get_tracks_by_genre("hip hop rap 2024",    "Hip Hop"))
all_tracks.extend(get_tracks_by_genre("indie rock 2024",     "Indie"))
all_tracks.extend(get_tracks_by_genre("pop hits 2023",       "Pop"))
all_tracks.extend(get_tracks_by_genre("hip hop rap 2023",    "Hip Hop"))
all_tracks.extend(get_tracks_by_genre("indie rock 2023",     "Indie"))

tracks_df = pd.DataFrame(all_tracks).drop_duplicates(subset='id')
print(f"Total tracks: {len(tracks_df)}")
tracks_df.head(10)
```

```python
# Cell 3 — Pull tracks via search (limit 10, more searches)
import pandas as pd

def get_tracks(search_term, genre_label):
    tracks = []
    results = sp.search(q=search_term, type='track', limit=10)
    for item in results['tracks']['items']:
        if item is None or item.get('id') is None:
            continue
        tracks.append({
            'id':     item['id'],
            'name':   item['name'],
            'artist': item['artists'][0]['name'],
            'genre':  genre_label
        })
    return tracks

all_tracks = []

# Pop — 4 searches x 10 = up to 40 tracks
for query in ["taylor swift", "dua lipa", "ariana grande", "billie eilish"]:
    all_tracks.extend(get_tracks(query, "Pop"))

# Hip Hop — 4 searches x 10 = up to 40 tracks
for query in ["kendrick lamar", "drake", "travis scott", "j cole"]:
    all_tracks.extend(get_tracks(query, "Hip Hop"))

# Indie — 4 searches x 10 = up to 40 tracks
for query in ["arctic monkeys", "tame impala", "the strokes", "phoebe bridgers"]:
    all_tracks.extend(get_tracks(query, "Indie"))

tracks_df = pd.DataFrame(all_tracks).drop_duplicates(subset='id')
print(f"Total tracks: {len(tracks_df)}")
tracks_df.head(10)
```

```python
# Cell 4 — Fetch audio features for all 120 tracks
def get_audio_features(track_ids):
    features = []
    for i in range(0, len(track_ids), 10):  # chunk by 10 to stay safe
        chunk = track_ids[i:i+10]
        results = sp.audio_features(chunk)
        features.extend([f for f in results if f is not None])
    return pd.DataFrame(features)

features_df = get_audio_features(tracks_df['id'].tolist())

# Merge song names + genre with audio features
df = tracks_df.merge(features_df, on='id')

print(f"Final dataset: {df.shape[0]} tracks, {df.shape[1]} columns")
df[['name', 'artist', 'genre', 'energy', 'danceability', 'valence', 'tempo']].head(10)
```

```python
# Cell 7 — Chart 1: Correlation heatmap
import seaborn as sns
import matplotlib.pyplot as plt

audio_cols = ['energy', 'danceability', 'valence', 'tempo',
              'acousticness', 'loudness', 'speechiness', 'instrumentalness']

plt.figure(figsize=(10, 8))
sns.heatmap(
    df[audio_cols].corr(),
    annot=True, fmt=".2f",
    cmap='coolwarm', center=0,
    square=True
)
plt.title("Audio Feature Correlations")
plt.tight_layout()
plt.savefig("chart1_correlations.png", dpi=150)
plt.show()
```

```python
# Cell 8 — Chart 2: Energy vs. Valence scatter by genre
plt.figure(figsize=(10, 6))
colors = {"pop": "#7F77DD", "hip-hop": "#1D9E75", "indie": "#EF9F27"}

for genre, group in df.groupby('genre'):
    plt.scatter(group['energy'], group['valence'],
                label=genre, alpha=0.6,
                color=colors[genre], s=60)

plt.xlabel("Energy")
plt.ylabel("Valence (happiness)")
plt.title("Energy vs. Happiness by Genre")
plt.legend()
plt.tight_layout()
plt.savefig("chart2_energy_valence.png", dpi=150)
plt.show()
```

```python
# Cell 9 — Chart 3: Danceability box plot
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df, x='genre', y='danceability',
    palette={"pop": "#AFA9EC", "hip-hop": "#5DCAA5", "indie": "#FAC775"}
)
plt.title("Danceability Distribution by Genre")
plt.tight_layout()
plt.savefig("chart3_danceability.png", dpi=150)
plt.show()
```

```python
# Cell 10 — Chart 4: Radar chart
import numpy as np

features_radar = ['energy', 'danceability', 'valence', 'acousticness', 'speechiness']
genre_means = df.groupby('genre')[features_radar].mean()

angles = np.linspace(0, 2 * np.pi, len(features_radar), endpoint=False).tolist()
angles += angles[:1]

fig, ax = plt.subplots(figsize=(7, 7), subplot_kw=dict(polar=True))

for genre, row in genre_means.iterrows():
    values = row.tolist() + row.tolist()[:1]
    ax.plot(angles, values, label=genre, linewidth=2)
    ax.fill(angles, values, alpha=0.1)

ax.set_xticks(angles[:-1])
ax.set_xticklabels(features_radar, size=12)
ax.set_title("Audio Profile by Genre", size=14, pad=20)
ax.legend(loc='upper right', bbox_to_anchor=(1.3, 1.1))
plt.tight_layout()
plt.savefig("chart4_radar.png", dpi=150)
plt.show()
```
